5.-  0.5 pto  
a) Hacer un diagrama de cajas para la variable tensión sistólica  
b) Hacer un diagrama de cajas para la variable tensión diastólica  
c) Existen valores superiores atípicos en ambos casos y a partir de que valor?  
d) Cuál es el IQR para la variable presión diastólica?

```python
#a) Hacer un diagrama de cajas para la variable tensión sistólica
plt.boxplot(eda['systolic'])
plt.title('Diagrama de cajas de tensión sistólica')
plt.ylabel('Presión')
plt.show()
#b) Hacer un diagrama de cajas para la variable tensión diastólica
plt.boxplot(eda['diastolic'])
plt.title('Diagrama de cajas de tensión diastólica')
plt.ylabel('Presión')
plt.show()
#c) Existen valores superiores atípicos en ambos casos y a partir de que valor?
q1 = eda['diastolic'].quantile(0.25)
q3 = eda['diastolic'].quantile(0.75)
iqr = q3 - q1
q1s = eda['systolic'].quantile(0.25)
q3s = eda['systolic'].quantile(0.75)
iqrs = q3s - q1s
print('Sí, en ambos diagramas se observan valores superiores atípicos:\nTensión diastólica: los valores atípicos comienzan aproximadamente por encima de', q3+1.5*iqr,'.\nTensión sistólica: los valores atípicos aparecen a partir de unos ',q3s+1.5*iqrs)
#d) Cuál es el IQR para la variable presión diastólica?
print('Rango intercuartilico:',iqr)
```

7.- 0.5 pto Existe correlación entre  las variables " estatura y presión sistólica" ?  

a)Si existe cuál es el valor de esta correlación ?  
b) Que conclusión podemos obtener de esta información ?

```python
corr = eda['height_cm'].corr(eda['systolic'])
print('Valor de la correlación:', corr)
```

b) Conclusión:

El valor de la correlación entre la estatura y la presión sistólica es aproximadamente {{corr}}. Un valor de correlación cercano a 0.21 indica una correlación positiva débil. Esto significa que hay una ligera tendencia a que las personas con mayor estatura tiendan a tener una presión sistólica ligeramente más alta, pero la relación no es fuerte.
